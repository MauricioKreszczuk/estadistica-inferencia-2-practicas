Investigar sobre la Paradoja de Simpson y comentar la relación que tiene con los modelos jerárquicos vistos en clase.



### 1. ¿Qué es la Paradoja de Simpson?

La **Paradoja de Simpson** es un fenómeno estadístico en el cual una tendencia, asociación o correlación observada dentro de múltiples subgrupos **desaparece o se invierte completamente** cuando los datos se analizan de manera agregada (combinando todos los grupos en uno solo).

Ocurre principalmente debido a la presencia de una **variable confusora o de agrupamiento** que afecta tanto a la variable predictora como a la respuesta, distribuyéndose de forma desigual entre los grupos.

---

### 2. Ejemplo Intuitivo: Tratamientos Médicos

Supongamos la comparación entre dos tratamientos ($A$ y $B$) para evaluar su tasa de éxito:

* **En casos leves:** El Tratamiento $A$ tiene una tasa de éxito del $90\%$, mientras que el $B$ alcanza el $80\%$.
* **En casos graves:** El Tratamiento $A$ tiene una tasa de éxito del $50\%$, mientras que el $B$ alcanza el $40\%$.
* **Datos combinados:** Al juntar todos los pacientes, el Tratamiento $B$ parece tener una tasa de éxito global superior al $A$.

**Explicación:** El tratamiento $A$ se administró principalmente a pacientes en estado grave (con baja probabilidad basal de supervivencia), mientras que el tratamiento $B$ se asignó mayoritariamente a pacientes leves. La severidad del caso actúa como la variable confusora que distorsiona la conclusión agregada.

---

### 3. Relación con los Modelos Jerárquicos

La Paradoja de Simpson es la consecuencia directa de **ignorar la estructura anidada o de agrupamiento de los datos**. Los modelos jerárquicos permiten abordar y resolver este problema al formalizar las distintas formas de agregar información:

#### A. Modelo con Agrupamiento Total (*Complete Pooling*)
* Asume un único parámetro global e ignora por completo la existencia de grupos.
* **Problema:** Es el más vulnerable a la Paradoja de Simpson, ya que confunde las diferencias entre grupos con el efecto real de la variable de interés.

#### B. Modelo sin Agrupamiento (*No Pooling*)
* Estima parámetros completamente independientes para cada grupo.
* **Problema:** Aunque evita la inversión de tendencias, genera estimaciones inestables y con alta varianza en grupos con tamaños de muestra pequeños, impidiendo extrapolar a la población general.

#### C. Modelo Jerárquico (*Partial Pooling*)
* Asume que los parámetros de cada grupo provienen de una distribución poblacional común gobernada por hiperparámetros ($\mu, \sigma$).
* **Resolución de la paradoja:** Permite descomponer simultáneamente:
  * **Efecto dentro del grupo (*within-group*):** Estima la relación real de las variables dentro de cada grupo específico.
  * **Efecto entre grupos (*between-group*):** Modela las diferencias estructurales y basales entre los distintos grupos.
* Al modelar pendientes e interceptos que varían por grupo mediante contracción hacia la media (*shrinkage*), el modelo jerárquico controla el efecto de la variable de agrupamiento y previene el sesgo de agregación.

## **Ejercicio 2:** Bayes Jerárquico

Escribir el teorema de Bayes para un modelo jerárquico.


En un modelo jerárquico se definen los siguientes componentes:

* $y = (y_1, y_2, \dots, y_J)$: Datos observados organizados en $J$ grupos.
* $\theta = (\theta_1, \theta_2, \dots, \theta_J)$: Parámetros específicos de cada grupo (nivel 2).
* $\psi = (\mu, \sigma, \dots)$: Hiperparámetros que gobiernan la distribución poblacional común de los grupos (nivel 3).

---

### Teorema de Bayes para la Distribución Posterior Conjunta

La distribución a posteriori conjunta de los parámetros específicos de grupo $\theta$ y los hiperparámetros $\psi$ dado los datos observados $y$ se formula como:

$$p(\theta, \psi \mid y) = \frac{p(y \mid \theta, \psi) \, p(\theta, \psi)}{p(y)}$$

Aplicando la regla de la cadena para factorizar la conjunta $p(\theta, \psi) = p(\theta \mid \psi) \, p(\psi)$ y considerando que los datos $y$ dependen de $\psi$ únicamente a través de $\theta$ (es decir, $p(y \mid \theta, \psi) = p(y \mid \theta)$):

$$p(\theta, \psi \mid y) = \frac{p(y \mid \theta) \, p(\theta \mid \psi) \, p(\psi)}{\int \int p(y \mid \theta) \, p(\theta \mid \psi) \, p(\psi) \, d\theta \, d\psi}$$

En forma proporcional (omitiendo la evidencia o constante de normalización):

$$p(\theta, \psi \mid y) \propto p(y \mid \theta) \, p(\theta \mid \psi) \, p(\psi)$$

---

### Descomposición por Niveles de Independencia Condicional

Bajo el supuesto de que las observaciones dentro de cada grupo son condicionalmente independientes dado su parámetro $\theta_j$, y que los parámetros $\theta_j$ son intercambiables condicionados a $\psi$:

$$p(\theta, \psi \mid y) \propto \left[ \prod_{j=1}^J p(y_j \mid \theta_j) \, p(\theta_j \mid \psi) \right] p(\psi)$$

Donde cada término corresponde a:

* $p(y_j \mid \theta_j)$: **Verosimilitud (Likelihood) - Nivel 1**, probabilidad de los datos del grupo $j$ dado su parámetro $\theta_j$.
* $p(\theta_j \mid \psi)$: **Prior de grupo - Nivel 2**, distribución poblacional común de los parámetros de cada grupo condicionada a los hiperparámetros.
* $p(\psi)$: **Hiperprior - Nivel 3**, distribución a priori de los parámetros globales de la población.

---

### Distribuciones Marginales Posteriores

* **Posterior Marginal de los Hiperparámetros ($\psi$):**
  Integrando respecto a los parámetros de grupo $\theta$:
  $$p(\psi \mid y) \propto p(\psi) \int p(y \mid \theta) \, p(\theta \mid \psi) \, d\theta = p(\psi) \, p(y \mid \psi)$$

* **Posterior Marginal de los Parámetros de Grupo ($\theta$):**
  Integrando respecto a los hiperparámetros $\psi$:
  $$p(\theta \mid y) = \int p(\theta, \psi \mid y) \, d\psi = \int p(\theta \mid \psi, y) \, p(\psi \mid y) \, d\psi$$

## **Ejercicio 3:**

Si tenemos los datos de atenciones mensuales de diferentes vacunatorios de la provincia. El objetivo es construir un modelo jerárquico para predecir las vacunaciones futuras y entender la variabilidad entre vacunatorios. Modelizar con un breve esquema de como podría ser un modelo de ese tipo. ¿Es jerárquico o no jerárquico el mejor planteo?


El **enfoque jerárquico (pooling parcial) es el mejor planteo**, debido a las siguientes razones:

1. **Heterogeneidad de tamaños muestrales:** Existen vacunatorios grandes (hospitales centrales) con alto volumen de atención y vacunatorios pequeños (salitas rurales) con pocas observaciones. Un modelo independiente (*no pooling*) sobreajustaría en los centros chicos, mientras que un modelo global (*complete pooling*) ignoraría las diferencias estructurales de cada centro.
2. **Compartición de información (*Shrinkage*):** Los centros con pocos registros o meses atípicos se benefician de la información poblacional global ($\mu, \sigma$), estabilizando las predicciones futuras.
3. **Cuantificación de la variabilidad ($\sigma$):** El hiperparámetro de dispersión permite responder directamente al objetivo de medir la heterogeneidad entre vacunatorios.
4. **Predicción para nuevos vacunatorios:** Permite hacer predicciones para un centro de salud nuevo del que no se tienen datos históricos, muestreando directamente desde la distribución poblacional $\mathcal{N}(\mu, \sigma)$.

---

### Esquema Matemático del Modelo (Regresión Poisson/Binomial Negativa)

Dado que las atenciones mensuales son **datos de conteo** ($y_{i,t} \in \mathbb{N}_0$ para el vacunatorio $i$ en el mes $t$):

#### Nivel 1: Datos observados (Likelihood)
$$y_{i,t} \sim \text{Poisson}(\lambda_{i,t}) \quad \text{o} \quad y_{i,t} \sim \text{NegativeBinomial}(\lambda_{i,t}, \alpha)$$

Como la tasa $\lambda_{i,t}$ debe ser estrictamente positiva, se utiliza la función de enlace logarítmica (*log-link*):

$$\log(\lambda_{i,t}) = \theta_i + \beta \cdot t$$
*(donde $t$ puede representar una tendencia temporal o estacionalidad y $\theta_i$ es el nivel basal del vacunatorio $i$)*

#### Nivel 2: Parámetros por Vacunatorio (Efectos de grupo)
$$\theta_i \sim \mathcal{N}(\mu, \sigma)$$

* $\theta_i$: Tasa basal esperada de atenciones del vacunatorio $i$ en escala logarítmica.

#### Nivel 3: Población de Vacunatorios (Hiperpriors)
$$\mu \sim \mathcal{N}(0, 2)$$
$$\sigma \sim \text{HalfNormal}(1)$$
$$\beta \sim \mathcal{N}(0, 1)$$

* $\mu$: Media global poblacional de atenciones en escala log.
* $\sigma$: Variabilidad o heterogeneidad entre los distintos vacunatorios de la provincia.

---

### 3. Diagrama Estructural de Niveles

```text
               (μ, σ, β) ~ p(ψ)                Nivel 3: Hiperpriors (Parámetros provinciales)
                       │
                       ▼
             θ_i ~ Normal(μ, σ)                Nivel 2: Efecto basal de cada vacunatorio i
                       │
                       ▼
       y_it ~ Poisson(exp(θ_i + β·t))          Nivel 1: Atenciones mensuales observadas
