Dados $\alpha$ y $\beta$, parámetros de una distribución Beta, interpreten el rol de ambos parámetros en términos de su PDF (probability density function).

Analizá en particular solo el caso de $\alpha >= 1$ y $\beta >= 1$**.

Recordar realizar la modelización completa, indicando cada componente de la distribución antes de realizar el análisis.

```python
'''
Si theta ~ Beta(alpha, beta)


theta ∈ [0,1]

alpha > 0 y beta > 0 

pdf = P(theta) = 1/B(alpha,beta) * theta**(alpha-1) * (1-theta)**(beta-1)

esperanza: E(theta) = alpha/(alpha + beta)

moda: Moda(theta) = (alpha-1)/(a+b-2)

varianza: Var(theta) = (alpha*beta)/((alpha + beta)**2 * (alpha + beta+ 1))
'''
```


### **3. Análisis del caso $\alpha \ge 1$ y $\beta \ge 1$**

Al restringir los parámetros a $\alpha \ge 1$ y $\beta \ge 1$, los exponentes del núcleo $(\alpha - 1 \ge 0$ y $\beta - 1 \ge 0)$ son no negativos. Esto elimina las asíntotas en los bordes ($\theta \to 0$ y $\theta \to 1$), garantizando densidades finitas en todo el soporte:

---

* **Caso 1: $\alpha = 1, \beta = 1$ (Uniforme)**
* Núcleo: $\theta^0 (1-\theta)^0 = 1$.
* La PDF es constante: $p(\theta) = 1$ para todo $\theta \in [0, 1]$.
* Representa un estado de **indiferencia total o ignorancia inicial**, donde todos los valores de $\theta$ son igualmente plausibles.
---

* **Caso 2: $\alpha = \beta > 1$ (Simétrica Unimodal)**
* Media y moda coinciden exactamente en $\mathbb{E}[\theta] = \text{Moda}(\theta) = 0.5$.
* La curva adopta forma de **campana simétrica** alrededor de $0.5$.
* Cuanto más grandes sean $\alpha$ y $\beta$ (por ejemplo, $\text{Beta}(20, 20)$ frente a $\text{Beta}(2, 2)$), más estrecha y alta es la campana, reflejando una fuerte creencia en que el proceso no está sesgado.
---

* **Caso 3: $\alpha > \beta \ge 1$ (Sesgo hacia el 1 / Derecha)**
* $\mathbb{E}[\theta] > 0.5$.
* Predominan los éxitos previos. La masa de probabilidad se desplaza hacia la derecha.
* Si $\beta = 1$ y $\alpha > 1$, la función es estrictamente creciente desde $0$ hasta su valor máximo en $\theta = 1$. Si $\beta > 1$, la moda queda en el interior del intervalo $(\frac{\alpha - 1}{\alpha + \beta - 2} > 0.5)$.
---

* **Caso 4: $\beta > \alpha \ge 1$ (Sesgo hacia el 0 / Izquierda)**
* $\mathbb{E}[\theta] < 0.5$.
* Predominan los fracasos previos. La masa de probabilidad se desplaza hacia la izquierda.
* Si $\alpha = 1$ y $\beta > 1$, la función es estrictamente decreciente con máximo en $\theta = 0$. Si $\alpha > 1$, la moda queda en el interior del intervalo $(\frac{\alpha - 1}{\alpha + \beta - 2} < 0.5)$.

## **Ejercicio 2:** un poco de historia.


a) El estadístico Bruno de Finetti declaró que "Las probabilidades no existen". Cómo se vincula este enunciado con el modelo BetaBinomial visto en clase?

b) Edwin Jaynes, físico, declaró que la teoría de probabilidad es la lógica de la ciencia. ¿Está de acuerdo con esta interpretación?


**a) Bruno de Finetti y "Las probabilidades no existen"**

* **Concepto:** De Finetti sostenía una postura subjetivista: la probabilidad no es una propiedad física de la moneda, sino el **grado de creencia e incertidumbre de un observador**.
* **Vínculo con el modelo:**
* El parámetro $\theta$ se modela como variable aleatoria con un prior $\text{Beta}(\alpha, \beta)$ precisamente porque refleja nuestra ignorancia/creencia previa, no un valor fijo de la naturaleza.
* Observar datos no "mide" una probabilidad real, sino que **actualiza de forma coherente el estado de conocimiento del observador** de un prior Beta a un posterior $\text{Beta}(\alpha + y, \beta + n - y)$.



---

**b) Edwin Jaynes y la probabilidad como "la lógica de la ciencia"**

* **Acuerdo / Fundamento:** Sí, porque la probabilidad bayesiana es una **extensión de la lógica deductiva tradicional** cuando hay información incompleta.
* **Vínculo con la ciencia:** El método científico opera cuantificando la plausibilidad de hipótesis (priors) frente a evidencia empírica (likelihood) para llegar a conclusiones consistentes (posteriors). La probabilidad matemática es el lenguaje formal de ese razonamiento.

## **Ejercicio 3:**
Sabemos que la media de una distribución Beta viene dada por:

$$\frac{α}{α+β}$$


Consideremos ahora un modelo Beta-Binomial con prior $Beta(3,4)$, y un experimento donde se realizaron 12 lanzamientos obteniendo 9 caras.

a) ¿Cuáles son los parámetros de la distribución a posteriori?

b) ¿Cuál es la media de la distribución posterior?

c) Comparar la media prior con la media posterior.

d)¿Cómo influyeron los datos observados sobre nuestras creencias iniciales?

```python
#a)
''' 
P(theta|Y) ~ Beta(alpha + y, beta+ n - y)

n = 12
y = 9
alpha = 3
beta = 4

alpha_post = alpha + y =  12
beta_post = beta + n - y = 7


#B
E(theta|Y) = alpha_post / (alpha_post+beta_post) = 12/19
'''
#c
'''
E(theta) = 3/7 =0.42
E(theta|Y) = 12/19 = 0.63
observada   theta_hat = 9/12 = 0.75

La media posterior resulta ser un promedio ponderado que se ubica estrictamente entre la creencia inicial ($0.4286$) y la evidencia observada ($0.7500$).

'''
```

**Datos iniciales:**

* **Prior:** $\text{Beta}(\alpha=3, \beta=4)$
* **Datos:** $n = 12$ lanzamientos, $y = 9$ caras (éxitos) $\implies n - y = 3$ cecas (fracasos)

---

**a) Parámetros de la distribución a posteriori**

Por conjugación Beta-Binomial:

* $\alpha_{\text{post}} = \alpha + y = 3 + 9 = 12$
* $\beta_{\text{post}} = \beta + n - y = 4 + 3 = 7$

$$\theta \mid y \sim \text{Beta}(12, 7)$$

---

**b) Media de la distribución posterior**

$$\mathbb{E}[\theta \mid y] = \frac{\alpha_{\text{post}}}{\alpha_{\text{post}} + \beta_{\text{post}}} = \frac{12}{12 + 7} = \frac{12}{19} \approx 0.6316$$

---

**c) Comparación entre la media prior y la media posterior**

* **Media a priori:** $\mathbb{E}[\theta] = \frac{3}{3 + 4} = \frac{3}{7} \approx 0.4286$
* **Proporción observada (datos):** $\hat{\theta} = \frac{y}{n} = \frac{9}{12} = 0.7500$
* **Media a posteriori:** $\mathbb{E}[\theta \mid y] = \frac{12}{19} \approx 0.6316$

La media posterior resulta ser un promedio ponderado que se ubica estrictamente entre la creencia inicial ($0.4286$) y la evidencia observada ($0.7500$).

---

**d) Influencia de los datos en las creencias iniciales**

* **Desplazamiento hacia la derecha:** Como se observó una alta proporción de caras ($75\%$), la evidencia empírica "tiró" la distribución hacia valores mayores a $0.5$, cambiando la creencia previa (que favorecía ligeramente a las cecas) a una que favorece claramente a las caras.
* **Reducción de incertidumbre:** El tamaño de muestra total equivalente pasó de $\kappa_{\text{prior}} = 7$ a $\kappa_{\text{post}} = 19$, haciendo que la distribución posterior sea más concentrada (menor varianza).

# **Ejercicio 4:**

Supongamos ahora un modelo Beta-Binomial con:

prior Beta(8,8), y 5 lanzamientos de moneda donde se observaron únicamente 1 cara.

a) Escribir la distribución a posteriori correspondiente.

b) Calcular la media posterior.

c) Comparar este resultado con la proporción observada $\frac{y}{N}$

  ¿Por qué el posterior no queda tan cerca de ¿$\frac{1}{5}$?
	​

Relacionar la respuesta con la influencia del prior y la cantidad de datos observados.

```python
'''
theta~ beta(8,8)

n = 5
y = 1

a)
P(theta|y) = theta**(alpha + y -1) * (1-theta)**(beta + n - y - 1) ~ beta(alpha_post, beta_post)
alpha_post = alpha + y = 9
beta_post = beta  + n - y = 12

-> theta|Y ~ beta(9,12)

b)
E(theta|Y) = alpha_post/(alpha_post + beta_post) = 9 / 21 = 3/7 = 0.42

c)
theta_hat = y/n = 1/5 = 0.2



'''
```

**Datos iniciales:**

* **Prior:** $\text{Beta}(\alpha=8, \beta=8)$
* **Datos:** $N = 5$ lanzamientos, $y = 1$ cara (éxito) $\implies N - y = 4$ cecas (fracasos)

---

**a) Distribución a posteriori**

Por conjugación Beta-Binomial:

* $\alpha_{\text{post}} = \alpha + y = 8 + 1 = 9$
* $\beta_{\text{post}} = \beta + N - y = 8 + 4 = 12$

$$\theta \mid y \sim \text{Beta}(9, 12)$$

---

**b) Media posterior**

$$\mathbb{E}[\theta \mid y] = \frac{\alpha_{\text{post}}}{\alpha_{\text{post}} + \beta_{\text{post}}} = \frac{9}{9 + 12} = \frac{9}{21} = \frac{3}{7} \approx 0.4286$$

---

**c) Comparación con la proporción observada y relación con la influencia del prior**

* **Proporción observada (datos):** $\hat{\theta} = \frac{y}{N} = \frac{1}{5} = 0.2000$
* **Media a priori:** $\mathbb{E}[\theta_{\text{prior}}] = \frac{8}{8+8} = 0.5000$
* **Media a posteriori:** $\mathbb{E}[\theta \mid y] = \frac{9}{21} \approx 0.4286$

**¿Por qué el posterior no queda tan cerca de $0.20$ ($1/5$)?**

1. **Peso relativo del prior (concentración previa):** El prior $\text{Beta}(8, 8)$ tiene una concentración de $\kappa_{\text{prior}} = \alpha + \beta = 16$, lo que equivale al peso de haber observado $16$ lanzamientos previos fuertemente centrados en $0.5$.
2. **Pocos datos observados:** La muestra solo aporta $N = 5$ observaciones. Como $16 > 5$, el prior tiene más del triple de peso relativo que la evidencia empírica en el cálculo del posterior:

$$\mathbb{E}[\theta \mid y] = \frac{16}{16 + 5}(0.50) + \frac{5}{16 + 5}(0.20) = \frac{16}{21}(0.50) + \frac{5}{21}(0.20) \approx 0.4286$$


3. **Efecto de regularización:** Como la muestra es muy pequeña frente a una creencia previa informativa y sólida, el posterior se mueve ligeramente hacia la izquierda (hacia el $0.20$), pero sigue fuertemente anclado cerca del valor $0.50$ impuesto por el prior.

## **Ejercicio 5:** referido al ejemplo de datos sintéticos obtienidos mediante un generador de números (pseudo)aleatorios, visto en la teoría.


1. Cambiar los valores de $N$ o $y$ para ver cómo cambia la concentración y el pico del posterior.

2. Cambiar los parámetros del prior para ver cómo influye la creencia previa (sugerencia: analizar qué sucede con un prior débil (curva más flexible y abierta) y  un prior fuerte (curva más concentrada alrededor de 0.5, incluso si los datos son pocos)).

3. Comparar distintos $N$ propuestos para entender qué sucede a medida que los datos aumentan (observen que todos los priors “convergen” hacia el mismo posterior).

```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D
from matplotlib.patches import Patch
import numpy as np
import preliz as pz

# 1. Simulación de datos sintéticos con valor real de theta
np.random.seed(42)
theta_real = 0.35
n_trials = [0, 1, 2, 4, 8, 16, 32, 64, 150, 500]

# Generación acumulada de éxitos (y) a partir de Bernoulli(theta_real)
tiradas = np.random.binomial(n=1, p=theta_real, size=n_trials[-1])
data = [0] + [tiradas[:N].sum() for N in n_trials[1:]]

# 2. Priors a comparar: No informativo, débilmente informativo y fuertemente informativo
beta_priors = {
    "No informativo: Beta(1, 1)": (1, 1),
    "Débil centrado: Beta(2, 2)": (2, 2),
    "Fuerte centrado: Beta(50, 50)": (50, 50),
}
colors = ["C0", "C1", "C2"]
x = np.linspace(0, 1, 1000)

# 3. Gráficos de evolución del posterior
fig, axes = plt.subplots(4, 3, figsize=(13, 11), sharex=True)
axes = axes.ravel()

for idx, N in enumerate(n_trials):
    ax = axes[idx]
    y = data[idx]

    for i, (name, (a_prior, b_prior)) in enumerate(beta_priors.items()):
        a_post = a_prior + y
        b_post = b_prior + N - y
        posterior = pz.Beta(a_post, b_post).pdf(x)
        ax.plot(x, posterior, color=colors[i], lw=2)
        ax.fill_between(x, 0, posterior, alpha=0.3, color=colors[i])

    ax.axvline(theta_real, color="k", linestyle="--", alpha=0.7, lw=1.5)
    ax.set_title(f"N = {N}, y = {y}", fontsize=10)
    ax.set_ylim(0, 25)
    ax.set_yticks([])

# Ocultar subplots sobrantes
for j in range(len(n_trials), len(axes)):
    fig.delaxes(axes[j])

# Leyenda global
legend_elements = [
    Patch(facecolor="C0", alpha=0.5, label="Beta(1, 1) [Plano/No informativo]"),
    Patch(facecolor="C1", alpha=0.5, label="Beta(2, 2) [Débil/Flexible]"),
    Patch(facecolor="C2", alpha=0.5, label="Beta(50, 50) [Fuerte/Informativo]"),
    Line2D([0], [0], color="k", linestyle="--", label=r"$\theta$ real (0.35)"),
]
fig.legend(
    handles=legend_elements,
    loc="lower right",
    bbox_to_anchor=(0.95, 0.08),
    fontsize=10,
)
plt.tight_layout()
plt.show()
```

## **Ejercicio 6:**

PreliZ tiene una función llamada [maxent](https://preliz.readthedocs.io/en/latest/api_reference.html#preliz.unidimensional.maxent). Explicar brevemente qué hace esta función y responder:

a)  Un investigador afirma que un parámetro debe ser positivo y que el 90% puede estar entre 2 y 20. Usá maxent para definir ese prior.

b) el mismo investigador afirma que un parámetro debe ser positivo con media 6 y 90% puede estar entre 2 y 20. Usar maxent para definir ese prior.




La función maxent busca los parámetros de una distribución de probabilidad que satisfagan un conjunto de restricciones dadas por el usuario (como intervalos de probabilidad, cuantiles, medias, etc.) bajo el Principio de Máxima Entropía (MaxEnt). Es decir, encuentra la distribución que cumple con la información provista introduciendo la menor cantidad de supuestos adicionales (la distribución menos sesgada o más conservadora).

a) Parámetro positivo con 90% de masa entre 2 y 20
Como el parámetro debe ser positivo, una elección estándar para modelarlo es una distribución Gamma o LogNormal. En maxent, definimos una instancia de la distribución y le pasamos el intervalo (2, 20) con masa 0.90 (usando mass=0.9 o cuantiles 0.05 y 0.95):

Porque Gamma?
* **Dominio:** Su soporte natural es $(0, \infty)$, garantizando que el parámetro sea estrictamente positivo.
* **Máxima entropía teórica:** Si se conoce la media de una variable positiva continua, la distribución de máxima entropía asociada es de la familia exponencial/Gamma.
* **Flexibilidad:** Sus dos parámetros de forma y escala permiten adaptarse tanto a comportamientos asimétricos sesgados a la derecha como a campanas concentradas sin permitir valores negativos.

```python
import preliz as pz

#definir la dist
gamma = pz.Gamma()

#aplicar maxent
pz.maxent(gamma, lower=2, upper=20, mass=0.9)

gamma.plot_pdf()
print(f"Prior a: {gamma}")
```

b) Parámetro positivo con media 6 y 90% de masa entre 2 y 20Al agregar la restricción de que la media debe ser $6$, le pasamos además el argumento mean=6

```python
import preliz as pz

gamma = pz.Gamma(mu=6)

# Usar mean=6 y mass=0.9
pz.maxent(gamma, lower=2, upper=20, mass=0.9)

# Ejecutar el método con ()
gamma.plot_pdf()
```

```python
import matplotlib.pyplot as plt
import preliz as pz

# 1. Definición y ajuste de priors
# a) 90% de masa entre 2 y 20
prior_a = pz.Gamma()
pz.maxent(prior_a, lower=2, upper=20, mass=0.9)

# b) Media fija = 6 y 90% de masa entre 2 y 20
prior_b = pz.Gamma(mu=6)
pz.maxent(prior_b, lower=2, upper=20, mass=0.9)

# 2. Gráfico comparativo
fig, axes = plt.subplots(1, 2, figsize=(11, 4), sharey=True)

prior_a.plot_pdf(ax=axes[0])
axes[0].set_title(f"a) 90% en [2, 20]\n{prior_a}")
axes[0].set_xlabel(r"$\theta$")
axes[0].set_ylabel("Densidad")

prior_b.plot_pdf(ax=axes[1])
axes[1].set_title(f"b) Media = 6 y 90% en [2, 20]\n{prior_b}")
axes[1].set_xlabel(r"$\theta$")

plt.tight_layout()
plt.show()
```

La idea de este ejercicio es aprender a **elicitar priors** (pasar conocimiento o intuición humana a una distribución matemática concreta) sin inventar números al azar:

* **No elegís hiperparámetros "a ojo":** En vez de adivinar qué valores de $\alpha$ y $\beta$ ponerle a una Gamma, le decís a la computadora lo que realmente sabés en el lenguaje del problema (*"está entre 2 y 20 con 90% de seguridad"*), y el algoritmo encuentra la matemática exacta.
* **Principio de Máxima Entropía (MaxEnt):** Aprendés a construir priors que cumplen estrictamente con lo que sabés, **sin agregar sesgos ni supuestos ocultos**. Es la distribución más "honesta" posible con la información que tenés.
* **El efecto de agregar información:** Al comparar **a** y **b**, ves que fijar la media en $6$ fuerza a la curva a reacomodarse (cambia su pico y cola) para respetar tanto el promedio como el rango de incertidumbre.
