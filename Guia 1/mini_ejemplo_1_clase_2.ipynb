```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta

# Valores de x
x = np.linspace(0, 1, 500)

# Configuraciones: (mu, kappa, etiqueta)
configuraciones = [
    (0.5, 40, r"$\mu=0.5,\ \kappa=40$"),
    (0.8, 40, r"$\mu=0.8,\ \kappa=40$"),
    (0.5, 5,  r"$\mu=0.5,\ \kappa=5$"),
    (0.5, 100, r"$\mu=0.5,\ \kappa=100$")
]

plt.figure(figsize=(10, 6))

for mu, kappa, etiqueta in configuraciones:

    alpha = mu * kappa
    beta_param = (1 - mu) * kappa

    y = beta.pdf(x, alpha, beta_param)

    plt.plot(
        x,
        y,
        linewidth=2.5,
        label=etiqueta
    )

plt.xlabel(r"$p$", fontsize=13)
plt.ylabel("Densidad", fontsize=13)
plt.title("Distribuciones Beta según μ y κ", fontsize=15)

plt.legend()
plt.grid(alpha=0.2)
plt.xlim(0, 1)

plt.show()
```

```python
mu = 0.5
kappa = 40

alpha = mu * kappa
beta_param = (1 - mu) * kappa

print(f"μ = {mu}")
print(f"κ = {kappa}")
print(f"α = {alpha}")
print(f"β = {beta_param}")
print(f"\nPrior: Beta({alpha:.0f}, {beta_param:.0f})")
```
